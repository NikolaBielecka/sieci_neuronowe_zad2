## Zadanie 2

W jaki sposób należy wykonać normalizację takich danych? **Napisz funkcję**, która na to pozwoli.


```python
def normalizacja(X):
    minimum = X.min(axis=0)
    maksimum = X.max(axis=0)
    X_norm = (X - minimum) / (maksimum - minimum)
    return X_norm
X_normalized = normalizacja(X)
print(X_normalized)
```

**Uzupełnij!**

Tutaj wyjaśnij z jakimi danymi mamy do czynienia i jak dokonać normalizacji (np. globalnie czy odrębnie dla każdej z cech, wyjaśnij dlaczego)
Dane posiadają 30 cech które mają różne zakresy wartości. Dlatego normalizację należy przeprowadzić osobno dla każdej cechy, dzięki temu wszystkie cechy będą miały wartości z przedziału od 0 do 1, bo cechyo większych wartościach mogłyby mieć większy wpływ na proces uczenia sieci neuronowej.

## Zadanie 3

Zapoznaj się z działaniem funkcji aktywacji `ReLU`. Spróbuj napisać własną wersję. Przetestuj na wygnerowanym wektorze danych za pomocą np. `numpy.random.rand()`

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

wektor = np.random.rand(10)

print("Wektor:")
print(wektor)

print("Po zastosowaniu ReLU:")
print(relu(wektor))
```

## Zadanie 4
Zapoznaj się z informacjami na temat pakietu Keras, a dokładniej na temat elementów które będziemy wykorzystywać na dzisiejszych zajęciach `tensorflow.keras.models.Sequential`, `tensorflow.keras.layers.Dense`, `tensorflow.keras.layers.Activation`, `tensorflow.keras.layers.Dropout`, `tensorflow.keras.utils.to_catgorical`

Uzuepłnij tekst poniżej oraz zaimportuj wyżej wymienione metody do swojego JupyterNotebook'a.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Activation
from tensorflow.keras.layers import Dropout
from tensorflow.keras.utils import to_categorical

```

**Uzupełnij:** 
- `Dense` służy do tworzenia warstwy neuronowej, jest to warstwa, która będzie uczona. Jej wartości będą się zmieniać
- `Activation` służy do okreslania funkcji aktywacji, jest to warstwa, która nie będzie uczona. 
- `Dropout` – służy do losowego wyłączania części neuronów podczas uczenia, aby zapobiegać przeuczeniu. Jest to warstwa nieucząca się.
- `Sequential` – służy do tworzenia modelu składającego się z kolejnych warstw.
- `to_categorical` – służy do przekształcania etykiet na postać one-hot.

## Zadanie 5
Stwórzmy model sieci za pomocą zaimportowanego wcześniej `tensorflow.keras.models.Sequential`.

```python
model = Sequential()
```

## Zadanie 6
Dodajmy warstwy do naszego modelu za pomocą `.add`. Dla pierwszej warstwy musimy określić `input_dim`, który będzie równy długości wektora cech przekazywanych na wejście naszej sieci.

Pierwsza warstwa niech będzie miała 20 neuronów, druga 10, oraz trzecia N neuronów. Ustal ile neuronów powinna mieć ostatnia warstwa na podstawie liczby klas analizowanego zbioru danych. Po pierwszej oraz drugiej z warstw neuronowych dodajmy "warstwę nieuczącą się" będącą funkcją aktywacji ReLU, oraz warstwę `Dropout` pozwalającą na regularyzację naszego modelu.

```python
model.add(Dense(20, input_dim=30))
model.add(Activation('relu'))
model.add(Dropout(0.2))

model.add(Dense(10))
model.add(Activation('relu'))
model.add(Dropout(0.2))

model.add(Dense(2))


```

Zbudujmy model (`.compile`), a następnie wyświetlmy o nim informacje (`.summary`).

## Zadanie 9
Wykorzystując `tensorflow.keras.utils.to_catgorical` stwórz etykiety `one-hot` i zapisz w zmiennej.

```python
labels_one_hot = to_categorical(y)

print(labels_one_hot)
```

## Zadanie 10
Zapoznaj się z działaniem metody `fit` dla stworzonego modelu. Jakie przyjmuje argumenty i co oznaczają?
Spróbuj nauczyć model z wykorzystaniem tej wymienionej metody oraz nieznormalizowanych danych - minibatch niech będzie wielkości 16, podział zbioru treningowego/testowego niech wynosi 8:2, liczba epok niech będzie równa 15. 

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    labels_one_hot,
    test_size=0.2
)
model.fit(
    X_train,
    y_train,
    batch_size=16,
    epochs=15
)
model.evaluate(X_test, y_test)


```
